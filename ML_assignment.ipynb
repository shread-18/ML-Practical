```python
import seaborn as sns

# Load the penguins dataset
penguins = sns.load_dataset('penguins')

# Display first 5 rows
penguins.head()
```

```python
print("Shape:", penguins.shape)
print("Rows:", penguins.shape[0])
print("Columns:", penguins.shape[1])
```

```python
penguins.info()
```

```python
penguins.describe(include='all')
```

```python
penguins.isnull().sum()
```

```python
print("Duplicate rows:", penguins.duplicated().sum())
```

```python
import matplotlib.pyplot as plt

penguins['species'].value_counts().plot(kind='bar')
plt.title("Species Distribution")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
import seaborn as sns

sns.pairplot(penguins, hue='species')
plt.show()
```

```python
numeric_data = penguins.select_dtypes(include='number')

plt.figure(figsize=(8,6))
sns.heatmap(numeric_data.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
numeric_columns = penguins.select_dtypes(include='number').columns

penguins[numeric_columns].hist(figsize=(10,8))
plt.tight_layout()
plt.show()
```

```python
for column in numeric_columns:
    plt.figure(figsize=(5,3))
    sns.boxplot(x=penguins[column])
    plt.title(f"Boxplot of {column}")
    plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='species', y='body_mass_g', data=penguins)
plt.title("Species vs Body Mass")
plt.show()
```

```python
sns.countplot(x='sex', hue='species', data=penguins)
plt.title("Sex Distribution by Species")
plt.show()
```

```python
sns.countplot(x='island', hue='species', data=penguins)
plt.title("Penguins on Different Islands")
plt.show()
```
